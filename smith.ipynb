```python
import skrf as rf
import matplotlib.pyplot as plt

# 1. Define the frequency range: 1 to 10 GHz with 101 points
freq = rf.Frequency(start=1, stop=10, npoints=101, unit='ghz')

# 2. Define the medium environment (system characteristic impedance Z0 = 50 Ohm)
media = rf.media.DefinedGammaZ0(frequency=freq, z0=50)

# 3. Create a component (e.g., a 100 Ohm shunt resistor)
shunt_resistor = media.resistor(100)

shunt_resistor
# shunt_resistor.plot_s_smith(m=1, n=0)

# plt.show()
```
