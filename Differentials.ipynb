```julia
H = sigmax()
ψ0 = basis(2, 0)
e_ops = [
    proj(basis(2, 0)),
    proj(basis(2, 1)),
    basis(2, 0) * basis(2, 1)'
]
tlist = LinRange(0, 10, 100)
sol = sesolve(H, ψ0, tlist, e_ops = e_ops, progress_bar = Val(false))

exp1 = real(sol.expect[1,:])
exp2 = real(sol.expect[2,:])
exp3 = real(sol.expect[3,:])

times = sol.times

plot!(times, exp1, label = "< 0 | p(t) | 0 >", xlabel = "t")
plot!(times, exp2, label = "< 1 | p(t) | 1 >")
plot!(times, exp3, label = "< 0 | p(t) | 0 >")
```
