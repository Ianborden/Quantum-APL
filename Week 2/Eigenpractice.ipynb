```julia
function random_density_matrix(n)
    A = randn(n, n) + im * randn(n, n)
    H = A + A'
    β = 1.0
    Identity = Matrix(I, n, n)
    while !isposdef(H)
        H = H + β * Identity
    end
    ρ = H/tr(H)
    return ρ
end

M = random_density_matrix(2)
eigen(M).values
```
