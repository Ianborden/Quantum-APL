```python
def S_21(f, Ql, Qc, fr):
    top = Ql/Qc
    bottom = 1+ 2j * Ql * (f/fr -1)
    return 1 - top/bottom

def Real(f,Ql,Qc,fr):
    R = 2j *Ql * (f/fr - 1)
    return (-R**2 - Ql/Qc + 1)/(1-R**2)

def Complex(f,Ql,Qc,fr):
    R = 2j *Ql * (f/fr - 1)
    return (2j * Ql**2/Qc * (f/fr - 1))/(1-R**2)
```
